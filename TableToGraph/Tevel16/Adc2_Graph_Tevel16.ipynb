```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://docs.google.com/spreadsheets/d/1SNtKn_p-X0sz3TAKy0b2ndkeiHQwIktb/export?format=csv&gid=1843564057"

df = pd.read_csv(url)

df.columns = df.columns.str.strip()

df['time'] = pd.to_datetime(df['time'], format="mixed", dayfirst=True)

df = df.sort_values('time')

df['smoothed'] = df['adc_conversion_radfet2'].rolling(window=20).mean()
plt.figure(figsize=(15,5))

plt.plot(df['time'], df['smoothed'])

plt.title('adc2 - time')
plt.xlabel('Time')
plt.ylabel('adc2')

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
