```python
import matplotlib.pyplot as plt
import pandas as pd

url = "https://docs.google.com/spreadsheets/d/1SNtKn_p-X0sz3TAKy0b2ndkeiHQwIktb/export?format=csv&gid=1843564057"

df = pd.read_csv(url)

df.columns = df.columns.str.strip()

df['time'] = pd.to_datetime(df['time'], format="mixed", dayfirst=True)

df = df.sort_values("time")

df['smoothed'] = df['temperature'].rolling(window=30).mean()
plt.figure(figsize=(15,5))

plt.plot(df["time"], df["smoothed"])

plt.xlabel("Time")
plt.ylabel("Temp")
plt.title("Temperature")

plt.xticks(rotation = 45)
plt.tight_layout()

plt.show()
```
