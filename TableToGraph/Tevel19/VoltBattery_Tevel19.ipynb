```python
import matplotlib.pyplot as plt
import pandas as pd

url = "https://docs.google.com/spreadsheets/d/1SNtKn_p-X0sz3TAKy0b2ndkeiHQwIktb/export?format=csv&gid=1692859273"

df = pd.read_csv(url)

df.columns = df.columns.str.strip()

df['sat_time'] = pd.to_datetime(df['sat_time'], format="mixed", dayfirst=True)

df = df.sort_values("sat_time")

df['smoothed'] = df['voltBattery'].rolling(window=5).mean()
plt.figure(figsize=(15,5))

plt.plot(df["sat_time"], df["smoothed"])

plt.xlabel("Time")
plt.ylabel("Volts")
plt.title("VoltBattery")

plt.xticks(rotation = 45)
plt.tight_layout()

plt.show()
```
