```python
import matplotlib.pyplot as plt
import pandas as pd

url = "https://docs.google.com/spreadsheets/d/1OVwNnElclqJ-Y4tWkePHjQtUetbhLSle/export?format=csv&gid=1865117878"

df = pd.read_csv(url)

df.columns = df.columns.str.strip()

df['time'] = pd.to_datetime(df['time'], format="mixed", dayfirst=True)

df = df.sort_values("time")

df['smoothed'] = df['adc_conversion_radfet1'].rolling(window=20).mean()
plt.figure(figsize=(15,5))

plt.plot(df["time"], df["smoothed"])

plt.xlabel("adc1")
plt.ylabel("Time")
plt.title("adc1 - time")

plt.xticks(rotation = 45)
plt.tight_layout()
```
