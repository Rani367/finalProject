```python
import matplotlib.pyplot as plt
import pandas as pd

url = "https://docs.google.com/spreadsheets/d/1OVwNnElclqJ-Y4tWkePHjQtUetbhLSle/export?format=csv&gid=791204570"

df = pd.read_csv(url)

df.columns = df.columns.str.strip()

df['time'] = pd.to_datetime(df['time'], format="mixed", dayfirst=True)

df = df.sort_values("time")

df['smoothed'] = df['SEU counter'].rolling(window=1).mean()
plt.figure(figsize=(15,5))

plt.plot(df["time"], df["smoothed"])

plt.xlabel("Time")
plt.ylabel("SEU counter")
plt.title("SEU")

plt.xticks(rotation = 45)
plt.tight_layout()
```
