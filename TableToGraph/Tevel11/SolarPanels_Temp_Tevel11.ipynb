```python
import matplotlib.pyplot as plt
import pandas as pd

url = "https://docs.google.com/spreadsheets/d/1OVwNnElclqJ-Y4tWkePHjQtUetbhLSle/export?format=csv&gid=1020539054"

df = pd.read_csv(url)

df.columns = df.columns.str.strip()

df['Ground Time'] = pd.to_datetime(df['Ground Time'], format="mixed", dayfirst=True)

df = df.sort_values("Ground Time")

plt.figure(figsize=(15,5))

# כאן פשוט מציירים 6 קווים (שני לפי שמות העמודות שלך)
plt.plot(df["Ground Time"], df["solar_panels0"].rolling(window=1000).mean(), label="temperature1")
plt.plot(df["Ground Time"], df["solar_panels1"].rolling(window=1000).mean(), label="temperature2")
plt.plot(df["Ground Time"], df["solar_panels2"].rolling(window=1000).mean(), label="temperature3")
plt.plot(df["Ground Time"], df["solar_panels3"].rolling(window=1000).mean(), label="temperature4")
plt.plot(df["Ground Time"], df["solar_panels4"].rolling(window=1000).mean(), label="temperature5")
plt.plot(df["Ground Time"], df["solar_panels5"].rolling(window=1000).mean(), label="temperature6")

plt.xlabel("Time")
plt.ylabel("Temp")
plt.title("Temp of solar panels")

plt.legend()  # מציג איזה קו זה מה

plt.xticks(rotation=45)
plt.tight_layout()
```
